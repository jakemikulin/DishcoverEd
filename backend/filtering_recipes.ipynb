```python
inverted_index = pkl.load(open("inverted_index_simple.pkl", "rb"))
```

```python
terms_counts = []
for key in inverted_index:
    positions = len(inverted_index[key])
    terms_counts.append((key, positions))
```

```python
terms_counts.sort(key=lambda x: x[1], reverse=True)
```

```python
import matplotlib.pyplot as plt

# plot zipf's law
x = range(1, len(terms_counts) + 1)
y = [count for term, count in terms_counts]
plt.plot(x, y)
plt.yscale('log')

plt.xlabel("Rank")
plt.ylabel("Frequency")
plt.show()
```

```python
top_5000_terms_no_count = [term for term, count in terms_counts[:5000]]
# Save list of top 5000 terms
pkl.dump(top_5000_terms_no_count, open("top_5000_terms.pkl", "wb"))
```
