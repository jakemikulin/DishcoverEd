```python
import pickle as pkl
```

```python
filename = "recipes_dict.pkl"
recipes_dict = pkl.load(open(filename, "rb"))
```

```python
filename = "inverted_index_simple.pkl"
inverted_index_simple = pkl.load(open
(filename, "rb"))

filename = "inverted_index_simple_titles.pkl"
inverted_index_simple_titles = pkl.load(open
(filename, "rb"))
```

```python
token_frequency = dict()
for token in inverted_index_simple:
    token_frequency[token] = len(inverted_index_simple[token])

for token in inverted_index_simple_titles:
    if token in token_frequency:
        token_frequency[token] += len(inverted_index_simple_titles[token])
    else:
        token_frequency[token] = len(inverted_index_simple_titles[token])

token_frequency
```

```python
# sorted list of tokens by frequency

sorted_tokens = sorted(token_frequency, key=token_frequency.get, reverse=True)
sorted_tokens = [(sorted_token, token_frequency[sorted_token]) for sorted_token in sorted_tokens]
sorted_tokens

# save sorted tokens as text file
with open('frequency_dict_recipes.txt', 'w') as f:
    for i in range(len(sorted_tokens)):
        sorted_token = sorted_tokens[i]
        f.write(str(sorted_token[0]) + " " + str(sorted_token[1]) + "\n")
```

```python
inverted_index = pkl.load(open("inverted_index_simple.pkl", "rb"))
```

```python
terms_counts = []
for key in inverted_index:
    positions = len(inverted_index[key])
    terms_counts.append((key, positions))
```

```python
terms_counts.sort(key=lambda x: x[1], reverse=True)
```

```python
import matplotlib.pyplot as plt

# plot zipf's law
x = range(1, len(terms_counts) + 1)
y = [count for term, count in terms_counts]
plt.plot(x, y)
plt.yscale('log')

plt.xlabel("Rank")
plt.ylabel("Frequency")
plt.show()
```

```python
i = 0
for term in terms_counts:
    if term[0] == "":
        print(i)
        break
    i+=1
```

```python
top_5000_terms_no_count = [term for term, count in terms_counts[:5000]]
# Save list of top 5000 terms
pkl.dump(top_5000_terms_no_count, open("top_5000_terms.pkl", "wb"))
```
